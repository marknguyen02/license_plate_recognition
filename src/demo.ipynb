```python
import cv2
import os
import matplotlib.pyplot as plt
from recognizer import PlateRecognizer
```

```python
project_path = os.path.dirname(os.getcwd())

yolo_ckpt = os.path.join(project_path, 'checkpoints', 'detect_best.pt')
digit_ckpt = os.path.join(project_path, 'checkpoints', 'digit_best.pth')
letter_ckpt = os.path.join(project_path, 'checkpoints', 'letter_best.pth')

input_names = [
    'carlong_0283.png',
    'carlong_0284.png'
]
input_paths = [ 
    os.path.join(project_path, 'data', 'images', fname) for fname in input_names
]
output_dir = os.path.join(project_path, 'tmp_outputs')

model = PlateRecognizer(
    yolo_ckpt=yolo_ckpt,
    digit_ckpt=digit_ckpt,
    letter_ckpt=letter_ckpt
)
```

```python
model.predict_batch(input_paths)
```

```python
input_imgs = [cv2.imread(path) for path in input_paths]

def show_img(img_cv2):
    img_rgb = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.axis("off")
    plt.show()

out_imgs = model.visualize_batch(input_imgs, output_dir=output_dir)
```

```python
for img in out_imgs:
    show_img(img)
```
